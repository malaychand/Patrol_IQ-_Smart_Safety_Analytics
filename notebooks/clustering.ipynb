```python
from google.colab import drive
drive.mount('/content/drive')

!pip install dagshub mlflow -q

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from sklearn.cluster import KMeans, DBSCAN, AgglomerativeClustering
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from scipy.cluster.hierarchy import dendrogram, linkage
import joblib
import json
import warnings
warnings.filterwarnings("ignore")

# MLflow and DagsHub setup
import dagshub
import mlflow
import mlflow.sklearn

BASE_PATH = "/content/drive/MyDrive/Malay_folder/guvi/"
RESULTS_PATH = BASE_PATH + "Patrool_results_set5/"
os.makedirs(RESULTS_PATH, exist_ok=True)

file_path = BASE_PATH + "Crimes_top_600k.csv"

df = pd.read_csv(file_path, low_memory=False)
print("Dataset shape:", df.shape)

df['Date'] = pd.to_datetime(df['Date'])
df["Hour"] = df["Date"].dt.hour
df["Month"] = df["Date"].dt.month
df["Year"] = df["Date"].dt.year

df = df.dropna(subset=['Latitude', 'Longitude']).reset_index(drop=True)

scaler = StandardScaler()
df[['Lat_norm', 'Lon_norm']] = scaler.fit_transform(
    df[['Latitude', 'Longitude']]
)

# Initialize DagsHub
dagshub.init(repo_owner='malaychand', repo_name='Patrol_IQ-_Smart_Safety_Analytics', mlflow=True)

# Configuration
SAMPLE_SIZE = 50000
HIERARCHICAL_SAMPLE = 5000
TEMPORAL_SAMPLE = 100000
RANDOM_STATE = 42

print("="*80)
print("PATROLIQ - CLUSTERING WITH ALL K VALUES")
print("="*80)

# =============================================================================
# PART 1: GEOGRAPHIC CRIME HOTSPOT CLUSTERING
# =============================================================================

print("\n" + "="*80)
print("PART 1: GEOGRAPHIC CRIME HOTSPOT CLUSTERING")
print("="*80)

# Prepare geographic features
geo_features = df[['Lat_norm', 'Lon_norm']].copy()
sample_geo = geo_features.sample(n=min(SAMPLE_SIZE, len(df)), random_state=RANDOM_STATE)

# -----------------------------------------------------------------------------
# 1.1 K-MEANS CLUSTERING - GENERATE ALL K VALUES (2-8)
# -----------------------------------------------------------------------------

print("\n" + "-"*80)
print("1.1 K-MEANS CLUSTERING - ALL K VALUES (2-8)")
print("-"*80)

with mlflow.start_run(run_name="Geographic_KMeans_All_K_Values"):

    # Test all K values
    K_range = range(2, 9)  # 2 to 8
    inertia_list = []
    silhouette_scores = []
    kmeans_models = {}

    print("\nTesting K values from 2 to 8...")

    for k in K_range:
        print(f"\n{'='*60}")
        print(f"Testing K={k}")
        print(f"{'='*60}")

        # Train KMeans
        kmeans = KMeans(n_clusters=k, random_state=RANDOM_STATE, n_init=10)
        labels = kmeans.fit_predict(sample_geo)

        # Calculate metrics
        inertia = kmeans.inertia_
        sil_score = silhouette_score(sample_geo, labels)
        db_score = davies_bouldin_score(sample_geo, labels)
        ch_score = calinski_harabasz_score(sample_geo, labels)

        inertia_list.append(inertia)
        silhouette_scores.append(sil_score)
        kmeans_models[k] = {
            'model': kmeans,
            'labels': labels,
            'silhouette': sil_score,
            'davies_bouldin': db_score,
            'calinski_harabasz': ch_score
        }

        print(f"K={k} Metrics:")
        print(f"  • Inertia: {inertia:.2f}")
        print(f"  • Silhouette Score: {sil_score:.4f}")
        print(f"  • Davies-Bouldin Index: {db_score:.4f}")
        print(f"  • Calinski-Harabasz Score: {ch_score:.2f}")

        # Cluster distribution
        cluster_counts = pd.Series(labels).value_counts().sort_index()
        print(f"\nCluster Distribution:")
        for cluster_id, count in cluster_counts.items():
            print(f"  Cluster {cluster_id}: {count:,} crimes ({count/len(labels)*100:.1f}%)")

        # Create visualization for this K
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 8))

        # Scatter plot
        scatter = ax1.scatter(sample_geo["Lon_norm"], sample_geo["Lat_norm"],
                             c=labels, cmap='tab10', s=10, alpha=0.6)
        centers = kmeans.cluster_centers_
        ax1.scatter(centers[:, 1], centers[:, 0],
                   c='red', marker='X', s=300, edgecolors='black', linewidths=2,
                   label='Cluster Centers')
        ax1.set_title(f"Geographic Crime Hotspots - KMeans (K={k})", fontsize=16, fontweight='bold')
        ax1.set_xlabel("Longitude (normalized)", fontsize=12)
        ax1.set_ylabel("Latitude (normalized)", fontsize=12)
        ax1.legend()
        ax1.grid(True, alpha=0.3)
        plt.colorbar(scatter, ax=ax1, label='Cluster ID')

        # Metrics display
        ax2.axis('off')
        metrics_text = f"""
        K-Means Clustering Results (K={k})

        Performance Metrics:
        ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
        Silhouette Score:        {sil_score:.4f}
        Davies-Bouldin Index:    {db_score:.4f}
        Calinski-Harabasz Score: {ch_score:.2f}
        Inertia:                 {inertia:.2f}

        Cluster Distribution:
        ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
        """
        for cluster_id, count in cluster_counts.items():
            metrics_text += f"\nCluster {cluster_id}: {count:,} ({count/len(labels)*100:.1f}%)"

        ax2.text(0.1, 0.5, metrics_text, fontsize=12, family='monospace',
                verticalalignment='center')

        plt.tight_layout()
        viz_path = RESULTS_PATH + f"geo_kmeans_k{k}_clusters.png"
        plt.savefig(viz_path, dpi=300, bbox_inches='tight')
        plt.show()

        # Save model
        model_path = RESULTS_PATH + f"kmeans_k{k}_model.pkl"
        joblib.dump(kmeans, model_path)

        # Log to MLflow
        mlflow.log_param(f"k_{k}_n_clusters", k)
        mlflow.log_metric(f"k_{k}_silhouette", sil_score)
        mlflow.log_metric(f"k_{k}_davies_bouldin", db_score)
        mlflow.log_metric(f"k_{k}_calinski_harabasz", ch_score)
        mlflow.log_artifact(viz_path)

        print(f"✓ Saved: {viz_path}")

    # Plot Elbow Curve with all K values
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

    ax1.plot(K_range, inertia_list, marker='o', linewidth=2, markersize=8)
    ax1.axvline(x=6, color='red', linestyle='--', linewidth=2, label='Recommended K=6')
    ax1.set_title("Elbow Method - All K Values", fontsize=14, fontweight='bold')
    ax1.set_xlabel("Number of Clusters (K)", fontsize=12)
    ax1.set_ylabel("Inertia", fontsize=12)
    ax1.legend()
    ax1.grid(True, alpha=0.3)
    ax1.set_xticks(K_range)

    ax2.plot(K_range, silhouette_scores, marker='s', color='green', linewidth=2, markersize=8)
    ax2.axvline(x=6, color='red', linestyle='--', linewidth=2, label='Recommended K=6')
    ax2.set_title("Silhouette Score - All K Values", fontsize=14, fontweight='bold')
    ax2.set_xlabel("Number of Clusters (K)", fontsize=12)
    ax2.set_ylabel("Silhouette Score", fontsize=12)
    ax2.legend()
    ax2.grid(True, alpha=0.3)
    ax2.set_xticks(K_range)

    plt.tight_layout()
    elbow_path = RESULTS_PATH + "geo_kmeans_elbow_all_k.png"
    plt.savefig(elbow_path, dpi=300, bbox_inches='tight')
    plt.show()

    mlflow.log_artifact(elbow_path)

    # Save comparison table
    kmeans_comparison = pd.DataFrame({
        'K': list(K_range),
        'Silhouette Score': silhouette_scores,
        'Inertia': inertia_list,
        'Davies-Bouldin': [kmeans_models[k]['davies_bouldin'] for k in K_range],
        'Calinski-Harabasz': [kmeans_models[k]['calinski_harabasz'] for k in K_range]
    })
    kmeans_comparison.to_csv(RESULTS_PATH + "kmeans_all_k_comparison.csv", index=False)

    print("\n" + "="*80)
    print("K-MEANS COMPARISON TABLE")
    print("="*80)
    print(kmeans_comparison.to_string(index=False))

    best_k_auto = kmeans_comparison.loc[kmeans_comparison['Silhouette Score'].idxmax(), 'K']
    print(f"\n✓ Best K by Silhouette Score: {best_k_auto}")

# -----------------------------------------------------------------------------
# 1.2 DBSCAN CLUSTERING - MULTIPLE EPS VALUES
# -----------------------------------------------------------------------------

print("\n" + "-"*80)
print("1.2 DBSCAN CLUSTERING - MULTIPLE EPS VALUES")
print("-"*80)

with mlflow.start_run(run_name="Geographic_DBSCAN_All_Eps"):

    eps_values = [0.05, 0.1, 0.15, 0.2, 0.25, 0.3, 0.35, 0.4]
    dbscan_models = {}

    print("\nTesting eps values:", eps_values)

    for eps in eps_values:
        print(f"\n{'='*60}")
        print(f"Testing eps={eps}")
        print(f"{'='*60}")

        # Train DBSCAN
        dbscan = DBSCAN(eps=eps, min_samples=20)
        labels = dbscan.fit_predict(sample_geo[['Lat_norm', 'Lon_norm']])

        # Calculate metrics
        n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
        n_noise = list(labels).count(-1)

        # Calculate silhouette only for valid clusters
        valid_idx = labels != -1
        if valid_idx.sum() > 1 and n_clusters > 1:
            sil_score = silhouette_score(sample_geo.loc[valid_idx, ['Lat_norm', 'Lon_norm']],
                                        labels[valid_idx])
            db_score = davies_bouldin_score(sample_geo.loc[valid_idx, ['Lat_norm', 'Lon_norm']],
                                           labels[valid_idx])
            ch_score = calinski_harabasz_score(sample_geo.loc[valid_idx, ['Lat_norm', 'Lon_norm']],
                                              labels[valid_idx])
        else:
            sil_score = db_score = ch_score = None

        dbscan_models[eps] = {
            'model': dbscan,
            'labels': labels,
            'n_clusters': n_clusters,
            'n_noise': n_noise,
            'silhouette': sil_score,
            'davies_bouldin': db_score,
            'calinski_harabasz': ch_score
        }

        print(f"eps={eps} Results:")
        print(f"  • Number of Clusters: {n_clusters}")
        print(f"  • Noise Points: {n_noise} ({n_noise/len(labels)*100:.1f}%)")
        if sil_score:
            print(f"  • Silhouette Score: {sil_score:.4f}")
            print(f"  • Davies-Bouldin Index: {db_score:.4f}")
            print(f"  • Calinski-Harabasz Score: {ch_score:.2f}")
        else:
            print(f"  • Metrics: N/A (insufficient clusters)")

        # Create visualization
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 8))

        # Scatter plot
        colors = labels.copy()
        scatter = ax1.scatter(sample_geo["Lon_norm"], sample_geo["Lat_norm"],
                             c=colors, cmap='tab10', s=10, alpha=0.6)
        ax1.set_title(f"Geographic Crime Hotspots - DBSCAN (eps={eps})",
                     fontsize=16, fontweight='bold')
        ax1.set_xlabel("Longitude (normalized)", fontsize=12)
        ax1.set_ylabel("Latitude (normalized)", fontsize=12)
        ax1.grid(True, alpha=0.3)
        plt.colorbar(scatter, ax=ax1, label='Cluster ID (-1 = Noise)')

        # Metrics display
        ax2.axis('off')
        metrics_text = f"""
        DBSCAN Clustering Results (eps={eps})

        Performance Metrics:
        ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
        Number of Clusters:      {n_clusters}
        Noise Points:            {n_noise} ({n_noise/len(labels)*100:.1f}%)
        """
        if sil_score:
            metrics_text += f"""
        Silhouette Score:        {sil_score:.4f}
        Davies-Bouldin Index:    {db_score:.4f}
        Calinski-Harabasz Score: {ch_score:.2f}
        """
        else:
            metrics_text += "\n        Metrics: N/A"

        # Cluster distribution
        if n_clusters > 0:
            metrics_text += "\n\n        Cluster Distribution:\n        ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━"
            cluster_counts = pd.Series(labels[labels != -1]).value_counts().sort_index()
            for cluster_id, count in cluster_counts.items():
                metrics_text += f"\n        Cluster {cluster_id}: {count:,} ({count/valid_idx.sum()*100:.1f}%)"

        ax2.text(0.1, 0.5, metrics_text, fontsize=12, family='monospace',
                verticalalignment='center')

        plt.tight_layout()
        viz_path = RESULTS_PATH + f"geo_dbscan_eps{eps:.2f}_clusters.png"
        plt.savefig(viz_path, dpi=300, bbox_inches='tight')
        plt.show()

        # Save model
        model_path = RESULTS_PATH + f"dbscan_eps{eps:.2f}_model.pkl"
        joblib.dump(dbscan, model_path)

        # Log to MLflow
        mlflow.log_param(f"eps_{eps}_value", eps)
        mlflow.log_metric(f"eps_{eps}_n_clusters", n_clusters)
        mlflow.log_metric(f"eps_{eps}_noise_points", n_noise)
        if sil_score:
            mlflow.log_metric(f"eps_{eps}_silhouette", sil_score)
        mlflow.log_artifact(viz_path)

        print(f"✓ Saved: {viz_path}")

    # Create comparison visualization
    eps_list = list(eps_values)
    n_clusters_list = [dbscan_models[e]['n_clusters'] for e in eps_list]
    n_noise_list = [dbscan_models[e]['n_noise'] for e in eps_list]
    sil_list = [dbscan_models[e]['silhouette'] if dbscan_models[e]['silhouette'] else 0 for e in eps_list]

    fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(18, 5))

    ax1.plot(eps_list, n_clusters_list, marker='o', linewidth=2, markersize=8, color='blue')
    ax1.set_title("Number of Clusters vs Eps", fontsize=14, fontweight='bold')
    ax1.set_xlabel("Epsilon (eps)", fontsize=12)
    ax1.set_ylabel("Number of Clusters", fontsize=12)
    ax1.grid(True, alpha=0.3)

    ax2.plot(eps_list, n_noise_list, marker='s', linewidth=2, markersize=8, color='red')
    ax2.set_title("Noise Points vs Eps", fontsize=14, fontweight='bold')
    ax2.set_xlabel("Epsilon (eps)", fontsize=12)
    ax2.set_ylabel("Number of Noise Points", fontsize=12)
    ax2.grid(True, alpha=0.3)

    ax3.plot(eps_list, sil_list, marker='^', linewidth=2, markersize=8, color='green')
    ax3.set_title("Silhouette Score vs Eps", fontsize=14, fontweight='bold')
    ax3.set_xlabel("Epsilon (eps)", fontsize=12)
    ax3.set_ylabel("Silhouette Score", fontsize=12)
    ax3.grid(True, alpha=0.3)

    plt.tight_layout()
    dbscan_comp_path = RESULTS_PATH + "geo_dbscan_eps_comparison.png"
    plt.savefig(dbscan_comp_path, dpi=300, bbox_inches='tight')
    plt.show()

    mlflow.log_artifact(dbscan_comp_path)

    # Save comparison table
    dbscan_comparison = pd.DataFrame({
        'Eps': eps_list,
        'N_Clusters': n_clusters_list,
        'Noise_Points': n_noise_list,
        'Silhouette Score': sil_list
    })
    dbscan_comparison.to_csv(RESULTS_PATH + "dbscan_all_eps_comparison.csv", index=False)

    print("\n" + "="*80)
    print("DBSCAN COMPARISON TABLE")
    print("="*80)
    print(dbscan_comparison.to_string(index=False))

# -----------------------------------------------------------------------------
# 1.3 HIERARCHICAL CLUSTERING - MULTIPLE N_CLUSTERS
# -----------------------------------------------------------------------------

print("\n" + "-"*80)
print("1.3 HIERARCHICAL CLUSTERING - MULTIPLE N_CLUSTERS")
print("-"*80)

with mlflow.start_run(run_name="Geographic_Hierarchical_All_N"):

    # Use smaller sample for hierarchical
    hier_sample = sample_geo[['Lat_norm', 'Lon_norm']].sample(
        n=min(HIERARCHICAL_SAMPLE, len(sample_geo)),
        random_state=RANDOM_STATE
    )

    print(f"\nUsing {len(hier_sample)} samples for Hierarchical Clustering...")

    n_clusters_range = range(2, 9)  # 2 to 8
    hier_models = {}

    print("\nTesting n_clusters from 2 to 8...")

    for n_clust in n_clusters_range:
        print(f"\n{'='*60}")
        print(f"Testing n_clusters={n_clust}")
        print(f"{'='*60}")

        # Train Hierarchical
        hier = AgglomerativeClustering(n_clusters=n_clust, linkage='ward')
        labels = hier.fit_predict(hier_sample)

        # Calculate metrics
        sil_score = silhouette_score(hier_sample, labels)
        db_score = davies_bouldin_score(hier_sample, labels)
        ch_score = calinski_harabasz_score(hier_sample, labels)

        hier_models[n_clust] = {
            'model': hier,
            'labels': labels,
            'silhouette': sil_score,
            'davies_bouldin': db_score,
            'calinski_harabasz': ch_score
        }

        print(f"n_clusters={n_clust} Metrics:")
        print(f"  • Silhouette Score: {sil_score:.4f}")
        print(f"  • Davies-Bouldin Index: {db_score:.4f}")
        print(f"  • Calinski-Harabasz Score: {ch_score:.2f}")

        # Cluster distribution
        cluster_counts = pd.Series(labels).value_counts().sort_index()
        print(f"\nCluster Distribution:")
        for cluster_id, count in cluster_counts.items():
            print(f"  Cluster {cluster_id}: {count:,} ({count/len(labels)*100:.1f}%)")

        # Create visualization
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 8))

        # Scatter plot
        scatter = ax1.scatter(hier_sample["Lon_norm"], hier_sample["Lat_norm"],
                             c=labels, cmap='tab10', s=20, alpha=0.7)
        ax1.set_title(f"Geographic Crime Hotspots - Hierarchical (n={n_clust})",
                     fontsize=16, fontweight='bold')
        ax1.set_xlabel("Longitude (normalized)", fontsize=12)
        ax1.set_ylabel("Latitude (normalized)", fontsize=12)
        ax1.grid(True, alpha=0.3)
        plt.colorbar(scatter, ax=ax1, label='Cluster ID')

        # Metrics display
        ax2.axis('off')
        metrics_text = f"""
        Hierarchical Clustering Results (n={n_clust})

        Performance Metrics:
        ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
        Silhouette Score:        {sil_score:.4f}
        Davies-Bouldin Index:    {db_score:.4f}
        Calinski-Harabasz Score: {ch_score:.2f}
        Linkage Method:          Ward

        Cluster Distribution:
        ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
        """
        for cluster_id, count in cluster_counts.items():
            metrics_text += f"\nCluster {cluster_id}: {count:,} ({count/len(labels)*100:.1f}%)"

        ax2.text(0.1, 0.5, metrics_text, fontsize=12, family='monospace',
                verticalalignment='center')

        plt.tight_layout()
        viz_path = RESULTS_PATH + f"geo_hierarchical_n{n_clust}_clusters.png"
        plt.savefig(viz_path, dpi=300, bbox_inches='tight')
        plt.show()

        # Save model
        model_path = RESULTS_PATH + f"hierarchical_n{n_clust}_model.pkl"
        joblib.dump(hier, model_path)

        # Log to MLflow
        mlflow.log_param(f"n_{n_clust}_clusters", n_clust)
        mlflow.log_metric(f"n_{n_clust}_silhouette", sil_score)
        mlflow.log_metric(f"n_{n_clust}_davies_bouldin", db_score)
        mlflow.log_metric(f"n_{n_clust}_calinski_harabasz", ch_score)
        mlflow.log_artifact(viz_path)

        print(f"✓ Saved: {viz_path}")

    # Generate dendrogram
    print("\nGenerating dendrogram...")
    linked = linkage(hier_sample, method='ward')

    plt.figure(figsize=(14, 7))
    dendrogram(linked, truncate_mode='level', p=5, color_threshold=0)
    plt.title("Hierarchical Clustering Dendrogram (Ward Linkage)",
             fontsize=16, fontweight='bold')
    plt.xlabel("Sample Index", fontsize=12)
    plt.ylabel("Distance", fontsize=12)
    plt.grid(True, alpha=0.3, axis='y')
    dendro_path = RESULTS_PATH + "geo_hierarchical_dendrogram.png"
    plt.savefig(dendro_path, dpi=300, bbox_inches='tight')
    plt.show()

    mlflow.log_artifact(dendro_path)

    # Comparison visualization
    n_list = list(n_clusters_range)
    sil_list = [hier_models[n]['silhouette'] for n in n_list]
    db_list = [hier_models[n]['davies_bouldin'] for n in n_list]
    ch_list = [hier_models[n]['calinski_harabasz'] for n in n_list]

    fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(18, 5))

    ax1.plot(n_list, sil_list, marker='o', linewidth=2, markersize=8, color='green')
    ax1.set_title("Silhouette Score vs N_Clusters", fontsize=14, fontweight='bold')
    ax1.set_xlabel("Number of Clusters", fontsize=12)
    ax1.set_ylabel("Silhouette Score", fontsize=12)
    ax1.grid(True, alpha=0.3)
    ax1.set_xticks(n_list)

    ax2.plot(n_list, db_list, marker='s', linewidth=2, markersize=8, color='orange')
    ax2.set_title("Davies-Bouldin Index vs N_Clusters", fontsize=14, fontweight='bold')
    ax2.set_xlabel("Number of Clusters", fontsize=12)
    ax2.set_ylabel("Davies-Bouldin Index", fontsize=12)
    ax2.grid(True, alpha=0.3)
    ax2.set_xticks(n_list)

    ax3.plot(n_list, ch_list, marker='^', linewidth=2, markersize=8, color='blue')
    ax3.set_title("Calinski-Harabasz Score vs N_Clusters", fontsize=14, fontweight='bold')
    ax3.set_xlabel("Number of Clusters", fontsize=12)
    ax3.set_ylabel("Calinski-Harabasz Score", fontsize=12)
    ax3.grid(True, alpha=0.3)
    ax3.set_xticks(n_list)

    plt.tight_layout()
    hier_comp_path = RESULTS_PATH + "geo_hierarchical_n_comparison.png"
    plt.savefig(hier_comp_path, dpi=300, bbox_inches='tight')
    plt.show()

    mlflow.log_artifact(hier_comp_path)

    # Save comparison table
    hier_comparison = pd.DataFrame({
        'N_Clusters': n_list,
        'Silhouette Score': sil_list,
        'Davies-Bouldin': db_list,
        'Calinski-Harabasz': ch_list
    })
    hier_comparison.to_csv(RESULTS_PATH + "hierarchical_all_n_comparison.csv", index=False)

    print("\n" + "="*80)
    print("HIERARCHICAL COMPARISON TABLE")
    print("="*80)
    print(hier_comparison.to_string(index=False))

    best_n_hier = hier_comparison.loc[hier_comparison['Silhouette Score'].idxmax(), 'N_Clusters']
    print(f"\n✓ Best n_clusters by Silhouette Score: {best_n_hier}")

# =============================================================================
# FINAL SUMMARY
# =============================================================================

print("\n" + "="*80)
print("CLUSTERING ANALYSIS COMPLETE!")
print("="*80)
print(f"\nAll results saved to: {RESULTS_PATH}")
print(f"\nGenerated Files:")
print(f"  • KMeans: 7 cluster visualizations (K=2 to K=8)")
print(f"  • DBSCAN: 8 eps visualizations")
print(f"  • Hierarchical: 7 cluster visualizations (n=2 to n=8)")
print(f"  • Comparison tables and charts for all algorithms")
print(f"\n{'='*80}\n")
```

```python
# =============================================================================
# PatrolIQ - Comprehensive EDA with Visualizations
# =============================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import warnings
import os
warnings.filterwarnings("ignore")

# Set style
plt.style.use("ggplot")
sns.set_palette("husl")

# Configuration
BASE_PATH = "/content/drive/MyDrive/Malay_folder/guvi/"
RESULTS_PATH = BASE_PATH + "Patrool_result_set2/"
os.makedirs(RESULTS_PATH, exist_ok=True)

print("="*80)
print("PATROLIQ - EXPLORATORY DATA ANALYSIS")
print("="*80)
print(f"\nResults will be saved to: {RESULTS_PATH}")

# Load data
file_path = BASE_PATH + "Crimes_top_600k.csv"
df = pd.read_csv(file_path, low_memory=False)

# Configure pandas display
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
pd.set_option('display.max_colwidth', 50)

print(f"\n✓ Data loaded successfully: {df.shape[0]:,} rows × {df.shape[1]} columns")

# =============================================================================
# PART 1: DATA OVERVIEW
# =============================================================================

print("\n" + "="*80)
print("PART 1: DATA OVERVIEW")
print("="*80)

# Basic information
print("\n1.1 Dataset Shape:")
print(f"   • Rows: {df.shape[0]:,}")
print(f"   • Columns: {df.shape[1]}")

# Data types
print("\n1.2 Data Types:")
print(df.dtypes.value_counts())

# Memory usage
memory_mb = df.memory_usage(deep=True).sum() / 1024**2
print(f"\n1.3 Memory Usage: {memory_mb:.2f} MB")

# Save data info
with open(RESULTS_PATH + "data_info.txt", "w") as f:
    f.write("PATROLIQ - DATASET INFORMATION\n")
    f.write("="*80 + "\n\n")
    f.write(f"Shape: {df.shape[0]:,} rows × {df.shape[1]} columns\n")
    f.write(f"Memory: {memory_mb:.2f} MB\n\n")
    f.write("Columns:\n")
    df.info(buf=f)

print(f"\n✓ Data info saved to: {RESULTS_PATH}data_info.txt")

# =============================================================================
# PART 2: MISSING VALUES ANALYSIS
# =============================================================================

print("\n" + "="*80)
print("PART 2: MISSING VALUES ANALYSIS")
print("="*80)

missing = df.isnull().sum().sort_values(ascending=False)
missing_pct = (missing / len(df)) * 100

missing_df = pd.DataFrame({
    "Missing_Count": missing,
    "Missing_Percent": missing_pct
})

missing_df = missing_df[missing_df["Missing_Count"] > 0]

if len(missing_df) > 0:
    print("\n2.1 Columns with Missing Values:")
    print(missing_df)

    # Visualize missing values
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

    # Bar chart
    missing_df.head(15).plot(kind='barh', y='Missing_Count', ax=ax1, color='coral')
    ax1.set_title('Missing Values Count (Top 15 Columns)', fontsize=14, fontweight='bold')
    ax1.set_xlabel('Missing Count', fontsize=12)
    ax1.set_ylabel('Column', fontsize=12)
    ax1.grid(True, alpha=0.3, axis='x')

    # Percentage chart
    missing_df.head(15).plot(kind='barh', y='Missing_Percent', ax=ax2, color='salmon')
    ax2.set_title('Missing Values Percentage (Top 15 Columns)', fontsize=14, fontweight='bold')
    ax2.set_xlabel('Missing Percentage (%)', fontsize=12)
    ax2.set_ylabel('Column', fontsize=12)
    ax2.grid(True, alpha=0.3, axis='x')

    plt.tight_layout()
    plt.savefig(RESULTS_PATH + "01_missing_values_analysis.png", dpi=300, bbox_inches='tight')
    plt.show()

    # Save to CSV
    missing_df.to_csv(RESULTS_PATH + "missing_values_summary.csv")
    print(f"\n✓ Missing values analysis saved")
else:
    print("\n✓ No missing values found!")

# =============================================================================
# PART 3: TEMPORAL FEATURE ENGINEERING
# =============================================================================

print("\n" + "="*80)
print("PART 3: TEMPORAL FEATURE ENGINEERING")
print("="*80)

# Convert dates
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
df['Updated On'] = pd.to_datetime(df['Updated On'], errors='coerce')

# Extract temporal features
df["Hour"] = df["Date"].dt.hour
df["Day"] = df["Date"].dt.day
df["Month"] = df["Date"].dt.month
df["Year"] = df["Date"].dt.year
df["DayOfWeek"] = df["Date"].dt.day_name()

# Season
def assign_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Spring"
    elif month in [6, 7, 8]:
        return "Summer"
    else:
        return "Fall"

df["Season"] = df["Month"].apply(assign_season)

# Weekend flag
df["Weekend"] = df["DayOfWeek"].isin(['Saturday', 'Sunday']).astype(int)

print("\n3.1 Temporal Features Created:")
print("   • Hour, Day, Month, Year")
print("   • DayOfWeek, Season, Weekend")

# Date range
print(f"\n3.2 Date Range:")
print(f"   • From: {df['Date'].min()}")
print(f"   • To: {df['Date'].max()}")
print(f"   • Span: {(df['Date'].max() - df['Date'].min()).days} days")

# =============================================================================
# PART 4: CRIME TYPE ANALYSIS
# =============================================================================

print("\n" + "="*80)
print("PART 4: CRIME TYPE ANALYSIS")
print("="*80)

crime_counts = df['Primary Type'].value_counts()

print(f"\n4.1 Total Crime Types: {df['Primary Type'].nunique()}")
print(f"\n4.2 Top 15 Crime Types:")
for i, (crime, count) in enumerate(crime_counts.head(15).items(), 1):
    pct = (count / len(df)) * 100
    print(f"   {i:2}. {crime:<30} {count:>7,} ({pct:>5.2f}%)")

# Visualize crime types
fig = plt.figure(figsize=(18, 10))

# Top 15 crimes - Bar chart
ax1 = plt.subplot(2, 2, 1)
top_15 = crime_counts.head(15)
colors = plt.cm.Reds(np.linspace(0.4, 0.9, 15))
bars = ax1.barh(range(len(top_15)), top_15.values, color=colors)
ax1.set_yticks(range(len(top_15)))
ax1.set_yticklabels(top_15.index)
ax1.set_xlabel('Number of Crimes', fontsize=11)
ax1.set_title('Top 15 Crime Types', fontsize=13, fontweight='bold')
ax1.grid(True, alpha=0.3, axis='x')
# Add value labels
for i, (bar, val) in enumerate(zip(bars, top_15.values)):
    ax1.text(val + 500, i, f'{val:,}', va='center', fontsize=9)

# Pie chart - Top 10
ax2 = plt.subplot(2, 2, 2)
top_10 = crime_counts.head(10)
others = crime_counts.iloc[10:].sum()
pie_data = pd.concat([top_10, pd.Series({'Others': others})])
colors_pie = plt.cm.Set3(np.linspace(0, 1, 11))
wedges, texts, autotexts = ax2.pie(pie_data.values, labels=pie_data.index, autopct='%1.1f%%',
                                     colors=colors_pie, startangle=90)
for autotext in autotexts:
    autotext.set_color('white')
    autotext.set_fontsize(9)
    autotext.set_fontweight('bold')
ax2.set_title('Crime Distribution (Top 10 + Others)', fontsize=13, fontweight='bold')

# Crime counts distribution
ax3 = plt.subplot(2, 2, 3)
ax3.hist(crime_counts.values, bins=30, color='steelblue', edgecolor='black', alpha=0.7)
ax3.set_xlabel('Number of Crimes', fontsize=11)
ax3.set_ylabel('Number of Crime Types', fontsize=11)
ax3.set_title('Distribution of Crime Frequencies', fontsize=13, fontweight='bold')
ax3.grid(True, alpha=0.3)

# Cumulative percentage
ax4 = plt.subplot(2, 2, 4)
cumsum = (crime_counts.cumsum() / crime_counts.sum() * 100)
ax4.plot(range(len(cumsum)), cumsum.values, linewidth=2, color='darkgreen')
ax4.axhline(y=80, color='red', linestyle='--', label='80% threshold')
ax4.fill_between(range(len(cumsum)), 0, cumsum.values, alpha=0.3, color='lightgreen')
ax4.set_xlabel('Crime Type Rank', fontsize=11)
ax4.set_ylabel('Cumulative Percentage (%)', fontsize=11)
ax4.set_title('Cumulative Crime Distribution', fontsize=13, fontweight='bold')
ax4.legend()
ax4.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig(RESULTS_PATH + "02_crime_type_analysis.png", dpi=300, bbox_inches='tight')
plt.show()

# Save crime counts
crime_counts.to_csv(RESULTS_PATH + "crime_type_counts.csv")
print(f"\n✓ Crime type analysis saved")

# =============================================================================
# PART 5: ARREST & DOMESTIC ANALYSIS
# =============================================================================

print("\n" + "="*80)
print("PART 5: ARREST & DOMESTIC ANALYSIS")
print("="*80)

# Arrest rate
arrest_rate = df['Arrest'].value_counts(normalize=True) * 100
print(f"\n5.1 Arrest Rate:")
print(f"   • Arrests Made: {arrest_rate.get(True, 0):.2f}%")
print(f"   • No Arrest: {arrest_rate.get(False, 0):.2f}%")

# Domestic incidents
domestic_rate = df['Domestic'].value_counts(normalize=True) * 100
print(f"\n5.2 Domestic Incidents:")
print(f"   • Domestic: {domestic_rate.get(True, 0):.2f}%")
print(f"   • Non-Domestic: {domestic_rate.get(False, 0):.2f}%")

# Arrest rates by crime type
arrest_by_crime = (
    df.groupby('Primary Type')['Arrest']
    .agg(
        total_crimes='size',
        arrests='sum',
        arrest_rate=lambda x: x.mean() * 100
    )
    .sort_values('arrest_rate', ascending=False)
    .round(2)
)

print(f"\n5.3 Top 10 Arrest Rates by Crime Type:")
print(arrest_by_crime.head(10))

# Visualize
fig = plt.figure(figsize=(18, 10))

# Arrest rate by crime type (top 15)
ax1 = plt.subplot(2, 2, 1)
top_arrest = arrest_by_crime.head(15)
colors = plt.cm.RdYlGn(np.linspace(0.3, 0.9, 15))
bars = ax1.barh(range(len(top_arrest)), top_arrest['arrest_rate'].values, color=colors)
ax1.set_yticks(range(len(top_arrest)))
ax1.set_yticklabels(top_arrest.index)
ax1.set_xlabel('Arrest Rate (%)', fontsize=11)
ax1.set_title('Top 15 Crime Types by Arrest Rate', fontsize=13, fontweight='bold')
ax1.grid(True, alpha=0.3, axis='x')
for i, (bar, val) in enumerate(zip(bars, top_arrest['arrest_rate'].values)):
    ax1.text(val + 1, i, f'{val:.1f}%', va='center', fontsize=9)

# Overall arrest distribution
ax2 = plt.subplot(2, 2, 2)
arrest_counts = df['Arrest'].value_counts()
colors_arrest = ['#ff6b6b', '#4ecdc4']
wedges, texts, autotexts = ax2.pie(arrest_counts.values,
                                     labels=['No Arrest', 'Arrest'],
                                     autopct='%1.1f%%',
                                     colors=colors_arrest, startangle=90)
for autotext in autotexts:
    autotext.set_color('white')
    autotext.set_fontsize(12)
    autotext.set_fontweight('bold')
ax2.set_title('Overall Arrest Rate', fontsize=13, fontweight='bold')

# Domestic vs Non-Domestic
ax3 = plt.subplot(2, 2, 3)
domestic_counts = df['Domestic'].value_counts()
colors_domestic = ['#45b7d1', '#f7dc6f']
wedges, texts, autotexts = ax3.pie(domestic_counts.values,
                                     labels=['Non-Domestic', 'Domestic'],
                                     autopct='%1.1f%%',
                                     colors=colors_domestic, startangle=90)
for autotext in autotexts:
    autotext.set_color('white')
    autotext.set_fontsize(12)
    autotext.set_fontweight('bold')
ax3.set_title('Domestic vs Non-Domestic Crimes', fontsize=13, fontweight='bold')

# Crime volume vs arrest rate scatter
ax4 = plt.subplot(2, 2, 4)
scatter_data = arrest_by_crime.head(20)
scatter = ax4.scatter(scatter_data['total_crimes'],
                     scatter_data['arrest_rate'],
                     s=scatter_data['arrests']/10,
                     alpha=0.6,
                     c=scatter_data['arrest_rate'],
                     cmap='RdYlGn')
ax4.set_xlabel('Total Crimes', fontsize=11)
ax4.set_ylabel('Arrest Rate (%)', fontsize=11)
ax4.set_title('Crime Volume vs Arrest Rate (Top 20)', fontsize=13, fontweight='bold')
ax4.grid(True, alpha=0.3)
plt.colorbar(scatter, ax=ax4, label='Arrest Rate (%)')

plt.tight_layout()
plt.savefig(RESULTS_PATH + "03_arrest_domestic_analysis.png", dpi=300, bbox_inches='tight')
plt.show()

# Save arrest analysis
arrest_by_crime.to_csv(RESULTS_PATH + "arrest_rates_by_crime.csv")
print(f"\n✓ Arrest analysis saved")

# =============================================================================
# PART 6: TEMPORAL PATTERNS (FIXED)
# =============================================================================

print("\n" + "="*80)
print("PART 6: TEMPORAL PATTERNS ANALYSIS")
print("="*80)

# Hourly patterns
hourly_counts = df['Hour'].value_counts().sort_index()
print(f"\n6.1 Peak Crime Hour: {hourly_counts.idxmax()}:00 ({hourly_counts.max():,} crimes)")

# Daily patterns
daily_counts = df['DayOfWeek'].value_counts()
print(f"\n6.2 Peak Crime Day: {daily_counts.idxmax()} ({daily_counts.max():,} crimes)")

# Monthly patterns
monthly_counts = df['Month'].value_counts().sort_index()
print(f"\n6.3 Peak Crime Month: {monthly_counts.idxmax()} ({monthly_counts.max():,} crimes)")

# Visualize temporal patterns
fig = plt.figure(figsize=(18, 12))

# Hourly distribution
ax1 = plt.subplot(3, 2, 1)
hours = range(24)
ax1.plot(hours, hourly_counts.values, linewidth=2, marker='o', markersize=6, color='#e74c3c')
ax1.fill_between(hours, hourly_counts.values, alpha=0.3, color='#e74c3c')
ax1.set_xticks(hours)
ax1.set_xlabel('Hour of Day', fontsize=11)
ax1.set_ylabel('Number of Crimes', fontsize=11)
ax1.set_title('Crime Distribution by Hour', fontsize=13, fontweight='bold')
ax1.grid(True, alpha=0.3)
# Mark peak hour
peak_hour = hourly_counts.idxmax()
ax1.axvline(x=peak_hour, color='red', linestyle='--', linewidth=2, label=f'Peak: {peak_hour}:00')
ax1.legend()

# Day of week
ax2 = plt.subplot(3, 2, 2)
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
day_counts = df['DayOfWeek'].value_counts().reindex(day_order)
colors_day = plt.cm.Blues(np.linspace(0.4, 0.9, 7))
bars = ax2.bar(range(7), day_counts.values, color=colors_day, edgecolor='black')
ax2.set_xticks(range(7))
ax2.set_xticklabels(day_order, rotation=45, ha='right')
ax2.set_ylabel('Number of Crimes', fontsize=11)
ax2.set_title('Crime Distribution by Day of Week', fontsize=13, fontweight='bold')
ax2.grid(True, alpha=0.3, axis='y')
for i, (bar, val) in enumerate(zip(bars, day_counts.values)):
    ax2.text(i, val + 500, f'{val:,}', ha='center', va='bottom', fontsize=9)

# Monthly distribution
ax3 = plt.subplot(3, 2, 3)
month_names = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
colors_month = plt.cm.Reds(np.linspace(0.4, 0.9, 12))
bars = ax3.bar(range(1, 13), monthly_counts.values, color=colors_month, edgecolor='black')
ax3.set_xticks(range(1, 13))
ax3.set_xticklabels(month_names, rotation=45, ha='right')
ax3.set_ylabel('Number of Crimes', fontsize=11)
ax3.set_title('Crime Distribution by Month', fontsize=13, fontweight='bold')
ax3.grid(True, alpha=0.3, axis='y')

# Seasonal distribution
ax4 = plt.subplot(3, 2, 4)
season_order = ['Winter', 'Spring', 'Summer', 'Fall']
season_counts = df['Season'].value_counts().reindex(season_order)
colors_season = ['#3498db', '#2ecc71', '#e74c3c', '#f39c12']
bars = ax4.bar(range(4), season_counts.values, color=colors_season, edgecolor='black')
ax4.set_xticks(range(4))
ax4.set_xticklabels(season_order)
ax4.set_ylabel('Number of Crimes', fontsize=11)
ax4.set_title('Crime Distribution by Season', fontsize=13, fontweight='bold')
ax4.grid(True, alpha=0.3, axis='y')
for i, (bar, val) in enumerate(zip(bars, season_counts.values)):
    ax4.text(i, val + 1000, f'{val:,}', ha='center', va='bottom', fontsize=10)

# Weekend vs Weekday
ax5 = plt.subplot(3, 2, 5)
weekend_counts = df['Weekend'].value_counts()
labels = ['Weekday', 'Weekend']
colors_weekend = ['#4ecdc4', '#ff6b6b']
bars = ax5.bar(range(2), [weekend_counts.get(0, 0), weekend_counts.get(1, 0)],
               color=colors_weekend, edgecolor='black', width=0.6)
ax5.set_xticks(range(2))
ax5.set_xticklabels(labels)
ax5.set_ylabel('Number of Crimes', fontsize=11)
ax5.set_title('Weekday vs Weekend Crimes', fontsize=13, fontweight='bold')
ax5.grid(True, alpha=0.3, axis='y')
for i, bar in enumerate(bars):
    height = bar.get_height()
    pct = (height / len(df)) * 100
    ax5.text(i, height + 1000, f'{height:,}\n({pct:.1f}%)',
            ha='center', va='bottom', fontsize=10, fontweight='bold')

# Heatmap: Hour x Day
ax6 = plt.subplot(3, 2, 6)
heatmap_data = df.groupby(['Hour', 'DayOfWeek']).size().reset_index(name='Count')
heatmap_pivot = heatmap_data.pivot(index='Hour', columns='DayOfWeek', values='Count').fillna(0)
heatmap_pivot = heatmap_pivot[day_order]
sns.heatmap(heatmap_pivot, annot=False, cmap='YlOrRd', ax=ax6, cbar_kws={'label': 'Crime Count'})
ax6.set_title('Crime Heatmap: Hour × Day', fontsize=13, fontweight='bold')
ax6.set_xlabel('Day of Week', fontsize=11)
ax6.set_ylabel('Hour of Day', fontsize=11)

plt.tight_layout()
plt.savefig(RESULTS_PATH + "04_temporal_patterns.png", dpi=300, bbox_inches='tight')
plt.show()

# FIXED: Save temporal data separately to avoid mixing index types
print("\nSaving temporal summaries...")

# Save hourly counts
hourly_df = pd.DataFrame({
    'Hour': hourly_counts.index,
    'Crime_Count': hourly_counts.values
})
hourly_df.to_csv(RESULTS_PATH + "temporal_hourly.csv", index=False)

# Save daily counts
daily_df = pd.DataFrame({
    'Day': day_counts.index,
    'Crime_Count': day_counts.values
})
daily_df.to_csv(RESULTS_PATH + "temporal_daily.csv", index=False)

# Save monthly counts
monthly_df = pd.DataFrame({
    'Month': monthly_counts.index,
    'Crime_Count': monthly_counts.values
})
monthly_df.to_csv(RESULTS_PATH + "temporal_monthly.csv", index=False)

# Save seasonal counts
seasonal_df = pd.DataFrame({
    'Season': season_counts.index,
    'Crime_Count': season_counts.values
})
seasonal_df.to_csv(RESULTS_PATH + "temporal_seasonal.csv", index=False)

# Create combined summary with separate rows
temporal_summary = pd.DataFrame({
    'Time_Period': ['Hour', 'Day', 'Month', 'Season'],
    'Peak_Value': [
        f"{hourly_counts.idxmax()}:00",
        daily_counts.idxmax(),
        str(monthly_counts.idxmax()),
        season_counts.idxmax()
    ],
    'Peak_Count': [
        hourly_counts.max(),
        daily_counts.max(),
        monthly_counts.max(),
        season_counts.max()
    ]
})
temporal_summary.to_csv(RESULTS_PATH + "temporal_summary.csv", index=False)

print(f"✓ Temporal patterns saved:")
print(f"  - temporal_hourly.csv")
print(f"  - temporal_daily.csv")
print(f"  - temporal_monthly.csv")
print(f"  - temporal_seasonal.csv")
print(f"  - temporal_summary.csv")
# =============================================================================
# PART 7: GEOGRAPHIC ANALYSIS
# =============================================================================

print("\n" + "="*80)
print("PART 7: GEOGRAPHIC ANALYSIS")
print("="*80)

# Coordinate ranges
print(f"\n7.1 Geographic Bounds:")
print(f"   • Latitude: {df['Latitude'].min():.4f} to {df['Latitude'].max():.4f}")
print(f"   • Longitude: {df['Longitude'].min():.4f} to {df['Longitude'].max():.4f}")

# District analysis
district_counts = df['District'].value_counts().sort_index()
print(f"\n7.2 Top 10 Districts by Crime Count:")
for i, (district, count) in enumerate(district_counts.head(10).items(), 1):
    print(f"   {i:2}. District {district:<3} {count:>7,} crimes")

# Visualize geographic patterns
fig = plt.figure(figsize=(18, 10))

# Geographic scatter (sampled)
ax1 = plt.subplot(2, 2, 1)
sample_geo = df.dropna(subset=['Latitude', 'Longitude']).sample(20000, random_state=42)
scatter = ax1.scatter(sample_geo['Longitude'], sample_geo['Latitude'],
                     s=1, alpha=0.3, c=sample_geo.index, cmap='viridis')
ax1.set_xlabel('Longitude', fontsize=11)
ax1.set_ylabel('Latitude', fontsize=11)
ax1.set_title('Geographic Crime Distribution (20k Sample)', fontsize=13, fontweight='bold')
ax1.grid(True, alpha=0.3)

# District bar chart
ax2 = plt.subplot(2, 2, 2)
top_districts = district_counts.head(15)
colors = plt.cm.Oranges(np.linspace(0.4, 0.9, 15))
bars = ax2.barh(range(len(top_districts)), top_districts.values, color=colors)
ax2.set_yticks(range(len(top_districts)))
ax2.set_yticklabels([f"District {d}" for d in top_districts.index])
ax2.set_xlabel('Number of Crimes', fontsize=11)
ax2.set_title('Top 15 Districts by Crime Count', fontsize=13, fontweight='bold')
ax2.grid(True, alpha=0.3, axis='x')

# Community Area
ax3 = plt.subplot(2, 2, 3)
community_counts = df['Community Area'].value_counts().head(15)
ax3.bar(range(len(community_counts)), community_counts.values,
        color='teal', edgecolor='black')
ax3.set_xticks(range(len(community_counts)))
ax3.set_xticklabels([f"CA {int(ca)}" for ca in community_counts.index], rotation=45, ha='right')
ax3.set_ylabel('Number of Crimes', fontsize=11)
ax3.set_title('Top 15 Community Areas', fontsize=13, fontweight='bold')
ax3.grid(True, alpha=0.3, axis='y')

# Location Description
ax4 = plt.subplot(2, 2, 4)
location_counts = df['Location Description'].value_counts().head(15)
colors = plt.cm.Purples(np.linspace(0.4, 0.9, 15))
bars = ax4.barh(range(len(location_counts)), location_counts.values, color=colors)
ax4.set_yticks(range(len(location_counts)))
ax4.set_yticklabels(location_counts.index, fontsize=9)
ax4.set_xlabel('Number of Crimes', fontsize=11)
ax4.set_title('Top 15 Crime Locations', fontsize=13, fontweight='bold')
ax4.grid(True, alpha=0.3, axis='x')

plt.tight_layout()
plt.savefig(RESULTS_PATH + "05_geographic_analysis.png", dpi=300, bbox_inches='tight')
plt.show()

# Save geographic data
district_counts.to_csv(RESULTS_PATH + "district_counts.csv")
print(f"\n✓ Geographic analysis saved")

# =============================================================================
# PART 8: CORRELATION ANALYSIS
# =============================================================================

print("\n" + "="*80)
print("PART 8: CORRELATION ANALYSIS")
print("="*80)

num_cols = ['Beat', 'District', 'Ward', 'Community Area', 'Latitude', 'Longitude',
            'Year', 'Month', 'Hour', 'Weekend', 'Arrest', 'Domestic']

# Filter existing columns
num_cols = [col for col in num_cols if col in df.columns]

corr = df[num_cols].corr()

print(f"\n8.1 Correlation Matrix ({len(num_cols)} features)")
print("\nTop 5 Positive Correlations:")
corr_flat = corr.abs().unstack()
corr_flat = corr_flat[corr_flat < 1]
top_corr = corr_flat.sort_values(ascending=False).head(5)
for (var1, var2), corr_val in top_corr.items():
    print(f"   • {var1} ↔ {var2}: {corr_val:.3f}")

# Visualize correlation
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 7))

# Heatmap
sns.heatmap(corr, annot=True, cmap='coolwarm', center=0, ax=ax1,
            square=True, linewidths=0.5, cbar_kws={"shrink": 0.8}, fmt='.2f')
ax1.set_title('Correlation Matrix', fontsize=14, fontweight='bold')

# Clustermap
corr_linkage = sns.clustermap(corr, cmap='coolwarm', center=0,
                               annot=True, fmt='.2f', figsize=(10, 10))
plt.close()  # Close the clustermap figure

# Plot top correlations
top_10_corr = corr_flat.sort_values(ascending=False).head(10)
ax2.barh(range(len(top_10_corr)), top_10_corr.values, color='steelblue', edgecolor='black')
ax2.set_yticks(range(len(top_10_corr)))
ax2.set_yticklabels([f"{v1} ↔ {v2}" for v1, v2 in top_10_corr.index], fontsize=9)
ax2.set_xlabel('Correlation Coefficient', fontsize=11)
ax2.set_title('Top 10 Feature Correlations', fontsize=13, fontweight='bold')
ax2.grid(True, alpha=0.3, axis='x')

plt.tight_layout()
plt.savefig(RESULTS_PATH + "06_correlation_analysis.png", dpi=300, bbox_inches='tight')
plt.show()

# Save correlation matrix
corr.to_csv(RESULTS_PATH + "correlation_matrix.csv")
print(f"\n✓ Correlation analysis saved")

# =============================================================================
# PART 9: CRIME DESCRIPTIONS
# =============================================================================

print("\n" + "="*80)
print("PART 9: CRIME DESCRIPTIONS ANALYSIS")
print("="*80)

print(f"\n9.1 Unique Descriptions: {df['Description'].nunique():,}")
print(f"\n9.2 Top 20 Crime Descriptions:")

top_desc = df['Description'].value_counts().head(20)
for i, (desc, count) in enumerate(top_desc.items(), 1):
    pct = (count / len(df)) * 100
    print(f"   {i:2}. {desc:<50} {count:>7,} ({pct:>5.2f}%)")

# Visualize descriptions
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 8))

# Bar chart
colors = plt.cm.Spectral(np.linspace(0, 1, 20))
bars = ax1.barh(range(len(top_desc)), top_desc.values, color=colors)
ax1.set_yticks(range(len(top_desc)))
ax1.set_yticklabels([desc[:40] + '...' if len(desc) > 40 else desc for desc in top_desc.index], fontsize=9)
ax1.set_xlabel('Number of Crimes', fontsize=11)
ax1.set_title('Top 20 Crime Descriptions', fontsize=13, fontweight='bold')
ax1.grid(True, alpha=0.3, axis='x')

# Word cloud data (top descriptions)
description_lengths = df['Description'].str.len()
ax2.hist(description_lengths, bins=50, color='coral', edgecolor='black', alpha=0.7)
ax2.set_xlabel('Description Length (characters)', fontsize=11)
ax2.set_ylabel('Frequency', fontsize=11)
ax2.set_title('Distribution of Description Lengths', fontsize=13, fontweight='bold')
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig(RESULTS_PATH + "07_descriptions_analysis.png", dpi=300, bbox_inches='tight')
plt.show()

# Save descriptions
top_desc.to_csv(RESULTS_PATH + "top_descriptions.csv")
print(f"\n✓ Descriptions analysis saved")

# =============================================================================
# PART 10: FEATURE ENGINEERING
# =============================================================================

print("\n" + "="*80)
print("PART 10: FEATURE ENGINEERING")
print("="*80)

from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.cluster import KMeans

# Check for missing values in coordinates
print(f"\n10.1 Data Quality Check:")
print(f"   • Missing Latitude: {df['Latitude'].isnull().sum():,}")
print(f"   • Missing Longitude: {df['Longitude'].isnull().sum():,}")

# Filter valid coordinates BEFORE any operations
df_valid = df.dropna(subset=['Latitude', 'Longitude']).copy()
print(f"   • Records with valid coordinates: {len(df_valid):,}")
print(f"   • Records dropped: {len(df) - len(df_valid):,}")

# Normalize coordinates (only for valid data)
scaler = StandardScaler()
df_valid[["Lat_norm", "Lon_norm"]] = scaler.fit_transform(
    df_valid[["Latitude", "Longitude"]]
)

# For original df, add normalized columns (with NaN where coordinates are missing)
df["Lat_norm"] = np.nan
df["Lon_norm"] = np.nan
df.loc[df_valid.index, "Lat_norm"] = df_valid["Lat_norm"]
df.loc[df_valid.index, "Lon_norm"] = df_valid["Lon_norm"]

# Binned coordinates (only for valid data)
df["Lat_bin"] = np.nan
df["Lon_bin"] = np.nan

df_valid["Lat_bin"] = pd.qcut(df_valid["Latitude"], 10, labels=False, duplicates='drop')
df_valid["Lon_bin"] = pd.qcut(df_valid["Longitude"], 10, labels=False, duplicates='drop')

df.loc[df_valid.index, "Lat_bin"] = df_valid["Lat_bin"]
df.loc[df_valid.index, "Lon_bin"] = df_valid["Lon_bin"]

# Geographic clustering (only for valid data)
print(f"\n10.2 Geographic Clustering...")
kmeans = KMeans(n_clusters=20, random_state=42, n_init='auto')
geo_clusters = kmeans.fit_predict(df_valid[["Latitude", "Longitude"]])

df["GeoCluster"] = np.nan
df.loc[df_valid.index, "GeoCluster"] = geo_clusters

print(f"   ✓ Created 20 geographic clusters")

# Crime severity
crime_severity = {
    "HOMICIDE": 5,
    "CRIM SEXUAL ASSAULT": 4,
    "ROBBERY": 4,
    "BATTERY": 3,
    "ASSAULT": 3,
    "BURGLARY": 3,
    "MOTOR VEHICLE THEFT": 3,
    "THEFT": 2,
    "CRIMINAL DAMAGE": 2,
    "DECEPTIVE PRACTICE": 2,
}
df["Severity"] = df["Primary Type"].map(crime_severity).fillna(1)

# Fill missing Location Description
df['Location Description'].fillna('UNKNOWN', inplace=True)

# Label encoding
label_crime = LabelEncoder()
label_location = LabelEncoder()

df["CrimeType_Encoded"] = label_crime.fit_transform(df["Primary Type"])
df["Location_Encoded"] = label_location.fit_transform(df["Location Description"])

print("\n10.3 Engineered Features:")
print("   • Lat_norm, Lon_norm (normalized coordinates)")
print("   • Lat_bin, Lon_bin (binned coordinates)")
print("   • GeoCluster (K=20 geographic clusters)")
print("   • Severity (1-5 scale)")
print("   • CrimeType_Encoded, Location_Encoded")
print("   • Season, Weekend (already created)")

print(f"\n✓ Total features now: {len(df.columns)}")

# Feature summary statistics
feature_summary = pd.DataFrame({
    'Feature': ['Lat_norm', 'Lon_norm', 'Lat_bin', 'Lon_bin', 'GeoCluster',
                'Severity', 'CrimeType_Encoded', 'Location_Encoded'],
    'Non_Null_Count': [
        df['Lat_norm'].notna().sum(),
        df['Lon_norm'].notna().sum(),
        df['Lat_bin'].notna().sum(),
        df['Lon_bin'].notna().sum(),
        df['GeoCluster'].notna().sum(),
        df['Severity'].notna().sum(),
        df['CrimeType_Encoded'].notna().sum(),
        df['Location_Encoded'].notna().sum()
    ],
    'Data_Type': [
        df['Lat_norm'].dtype,
        df['Lon_norm'].dtype,
        df['Lat_bin'].dtype,
        df['Lon_bin'].dtype,
        df['GeoCluster'].dtype,
        df['Severity'].dtype,
        df['CrimeType_Encoded'].dtype,
        df['Location_Encoded'].dtype
    ]
})

print(f"\n10.4 Feature Summary:")
print(feature_summary.to_string(index=False))

# Visualize severity distribution
fig = plt.figure(figsize=(18, 10))

# Severity distribution
ax1 = plt.subplot(2, 3, 1)
severity_counts = df['Severity'].value_counts().sort_index()
colors_sev = plt.cm.YlOrRd(np.linspace(0.3, 0.9, len(severity_counts)))
bars = ax1.bar(severity_counts.index, severity_counts.values, color=colors_sev, edgecolor='black')
ax1.set_xlabel('Severity Level', fontsize=11)
ax1.set_ylabel('Number of Crimes', fontsize=11)
ax1.set_title('Crime Severity Distribution', fontsize=13, fontweight='bold')
ax1.grid(True, alpha=0.3, axis='y')
for bar, val in zip(bars, severity_counts.values):
    ax1.text(bar.get_x() + bar.get_width()/2, val + 1000, f'{val:,}',
            ha='center', va='bottom', fontsize=10)

# Geographic clusters (only valid data)
ax2 = plt.subplot(2, 3, 2)
geo_cluster_counts = df['GeoCluster'].value_counts().sort_index()
ax2.bar(geo_cluster_counts.index, geo_cluster_counts.values, color='teal', edgecolor='black')
ax2.set_xlabel('Geographic Cluster ID', fontsize=11)
ax2.set_ylabel('Number of Crimes', fontsize=11)
ax2.set_title('Geographic Cluster Distribution (K=20)', fontsize=13, fontweight='bold')
ax2.grid(True, alpha=0.3, axis='y')

# Normalized coordinates distribution
ax3 = plt.subplot(2, 3, 3)
ax3.hist2d(df_valid['Lon_norm'], df_valid['Lat_norm'], bins=50, cmap='hot')
ax3.set_xlabel('Longitude (normalized)', fontsize=11)
ax3.set_ylabel('Latitude (normalized)', fontsize=11)
ax3.set_title('Normalized Coordinate Heatmap', fontsize=13, fontweight='bold')
plt.colorbar(ax3.collections[0], ax=ax3, label='Crime Density')

# Binned coordinates
ax4 = plt.subplot(2, 3, 4)
lat_bin_counts = df['Lat_bin'].value_counts().sort_index()
ax4.bar(lat_bin_counts.index, lat_bin_counts.values, color='steelblue', edgecolor='black')
ax4.set_xlabel('Latitude Bin', fontsize=11)
ax4.set_ylabel('Number of Crimes', fontsize=11)
ax4.set_title('Latitude Bin Distribution', fontsize=13, fontweight='bold')
ax4.grid(True, alpha=0.3, axis='y')

ax5 = plt.subplot(2, 3, 5)
lon_bin_counts = df['Lon_bin'].value_counts().sort_index()
ax5.bar(lon_bin_counts.index, lon_bin_counts.values, color='coral', edgecolor='black')
ax5.set_xlabel('Longitude Bin', fontsize=11)
ax5.set_ylabel('Number of Crimes', fontsize=11)
ax5.set_title('Longitude Bin Distribution', fontsize=13, fontweight='bold')
ax5.grid(True, alpha=0.3, axis='y')

# Crime type encoding
ax6 = plt.subplot(2, 3, 6)
crime_encoded_counts = df['CrimeType_Encoded'].value_counts().head(15).sort_index()
ax6.bar(crime_encoded_counts.index, crime_encoded_counts.values, color='purple', edgecolor='black')
ax6.set_xlabel('Encoded Crime Type', fontsize=11)
ax6.set_ylabel('Number of Crimes', fontsize=11)
ax6.set_title('Crime Type Encoding (Top 15)', fontsize=13, fontweight='bold')
ax6.grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.savefig(RESULTS_PATH + "08_feature_engineering.png", dpi=300, bbox_inches='tight')
plt.show()

# Save feature summary
feature_summary.to_csv(RESULTS_PATH + "feature_summary.csv", index=False)

# Save label encodings for reference
crime_type_mapping = pd.DataFrame({
    'Crime_Type': label_crime.classes_,
    'Encoded_Value': range(len(label_crime.classes_))
})
crime_type_mapping.to_csv(RESULTS_PATH + "crime_type_encoding.csv", index=False)

location_mapping = pd.DataFrame({
    'Location_Description': label_location.classes_,
    'Encoded_Value': range(len(label_location.classes_))
})
location_mapping.to_csv(RESULTS_PATH + "location_encoding.csv", index=False)

# Save severity mapping
severity_mapping = pd.DataFrame({
    'Crime_Type': list(crime_severity.keys()) + ['Others'],
    'Severity_Level': list(crime_severity.values()) + [1]
})
severity_mapping.to_csv(RESULTS_PATH + "severity_mapping.csv", index=False)

print(f"\n✓ Feature engineering visualizations saved")
print(f"✓ Feature mappings saved:")
print(f"   - crime_type_encoding.csv")
print(f"   - location_encoding.csv")
print(f"   - severity_mapping.csv")

# Save engineered dataset
output_path = RESULTS_PATH + "feature_engineered_crimes.csv"
df.to_csv(output_path, index=False)
print(f"\n✓ Feature-engineered dataset saved to: {output_path}")
print(f"   • Total records: {len(df):,}")
print(f"   • Total features: {len(df.columns)}")
print(f"   • Records with valid geo data: {df['GeoCluster'].notna().sum():,}")

# =============================================================================
# PART 11: FINAL SUMMARY
# =============================================================================

print("\n" + "="*80)
print("PART 11: EDA SUMMARY")
print("="*80)

summary_stats = {
    "Total Records": len(df),
    "Total Columns": len(df.columns),
    "Date Range": f"{df['Date'].min()} to {df['Date'].max()}",
    "Crime Types": df['Primary Type'].nunique(),
    "Districts": df['District'].nunique(),
    "Arrest Rate": f"{(df['Arrest'].sum() / len(df) * 100):.2f}%",
    "Domestic Rate": f"{(df['Domestic'].sum() / len(df) * 100):.2f}%",
    "Peak Hour": f"{hourly_counts.idxmax()}:00",
    "Peak Day": daily_counts.idxmax(),
    "Peak Month": monthly_counts.idxmax(),
    "Most Common Crime": crime_counts.idxmax(),
    "Top District": f"District {district_counts.idxmax()}"
}

print("\nKey Statistics:")
for key, value in summary_stats.items():
    print(f"   • {key}: {value}")

# Save summary
with open(RESULTS_PATH + "eda_summary.txt", "w") as f:
    f.write("PATROLIQ - EDA SUMMARY\n")
    f.write("="*80 + "\n\n")
    for key, value in summary_stats.items():
        f.write(f"{key}: {value}\n")

print(f"\n{'='*80}")
print("EDA COMPLETE!")
print(f"{'='*80}")
print(f"\n✓ All results saved to: {RESULTS_PATH}")
print(f"\nGenerated Files:")
print("   1. data_info.txt")
print("   2. 01_missing_values_analysis.png")
print("   3. 02_crime_type_analysis.png")
print("   4. 03_arrest_domestic_analysis.png")
print("   5. 04_temporal_patterns.png")
print("   6. 05_geographic_analysis.png")
print("   7. 06_correlation_analysis.png")
print("   8. 07_descriptions_analysis.png")
print("   9. 08_feature_engineering.png")
print("   10. feature_engineered_crimes.csv")
print("   11. eda_summary.txt")
print("   + Various CSV summaries")

print(f"\n{'='*80}\n")
```
