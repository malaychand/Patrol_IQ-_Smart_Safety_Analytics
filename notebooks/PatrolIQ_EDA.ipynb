## Feature engineering for clustering

```python
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.cluster import KMeans
```

```python
def assign_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Spring"
    elif month in [6, 7, 8]:
        return "Summer"
    else:
        return "Fall"

df["Season"] = df["Month"].apply(assign_season)
```

```python
scaler = StandardScaler()
df[["Lat_norm", "Lon_norm"]] = scaler.fit_transform(df[["Latitude", "Longitude"]])
```

```python
df["Lat_bin"] = pd.qcut(df["Latitude"], 10, labels=False)
df["Lon_bin"] = pd.qcut(df["Longitude"], 10, labels=False)
```

```python
kmeans = KMeans(n_clusters=20, random_state=42, n_init='auto')
df["GeoCluster"] = kmeans.fit_predict(df[["Latitude", "Longitude"]])
```

```python
crime_severity = {
    "HOMICIDE": 5,
    "CRIM SEXUAL ASSAULT": 4,
    "ROBBERY": 4,
    "BATTERY": 3,
    "ASSAULT": 3,
    "BURGLARY": 3,
    "MOTOR VEHICLE THEFT": 3,
    "THEFT": 2,
    "CRIMINAL DAMAGE": 2,
    "DECEPTIVE PRACTICE": 2,
}
```

```python
df["Severity"] = df["Primary Type"].map(crime_severity).fillna(1)
```

```python
label = LabelEncoder()

df["CrimeType_Encoded"] = label.fit_transform(df["Primary Type"])
df["Location_Encoded"] = label.fit_transform(df["Location Description"])
```
